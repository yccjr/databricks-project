```python
df = spark.read.format("csv").option("header", "true").option("inferSchema", "true").load("/databricks-datasets/Rdatasets/data-001/csv/ggplot2/diamonds.csv")


display(df)
```

```python
##creates a temporary SQL table - 'diamonds_table'
df.createOrReplaceTempView("diamonds_table")
```

```python
%sql
-- avg price and avg carat size for each diamond cut
select
  cut, 
  ROUND(AVG(price), 2) AS average_price, 
  ROUND(AVG(carat), 2) AS average_carat
FROM   diamonds_table
GROUP BY 1 
ORDER BY 2 desc; 
```
