```python
%sql
create or replace view default.marketing_gold as
select  cut as  campaign_tier
    , carat as budget_invested_usd
    , price  as total_sales_usd
  -- successful purchases= price > $5000)
    , case when price > 5000 then 1 else 0 end as successful_purchases  
  -- roas 
    , round(price / carat, 2) AS return_on_ad_spend_ratio
    , case when  (price / carat) >= 5000 THEN 'High Return' else 'Standard Return' end as performance_efficiency_flag
from   read_files(
    '/databricks-datasets/Rdatasets/data-001/csv/ggplot2/diamonds.csv',
    format => 'csv',
    header => true
  );

COMMENT ON VIEW default.marketing_gold IS 
'The official marketing semantic layer. Use this to track budget, sales performance, and campaign Return on Ad Spend (ROAS).';
```

```python
%sql 
--testing
select   campaign_tier
         , round(SUM(budget_invested_usd), 2) AS total_budget_invested
         , avg(return_on_ad_spend_ratio) AS average_roas
from    default.marketing_gold
group by  1  
order by 2  desc  
```
